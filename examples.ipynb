```python
import ograph.ofig as of
import ograph.oplt as op
import matplotlib.pyplot as plt
import numpy as np
```

Create an Axes3D

```python
of.fig3()
op.annotate("Title", "x", "y", "z")

plt.show()
```

Plot lines, functions, and scatter points.

```python
fig = plt.figure()
ax = plt.axes(projection='3d')

# Plot a line that connects [0,0,0] and [0,0,10*np.pi]
ax.plot3D([0,0], [0,0], [0,10*np.pi])

# Plot a spiral
t = np.arange(0, 10*np.pi, np.pi/50)
x = np.sin(t)
y = np.cos(t)
ax.plot3D(x, y, t)

# Plot scatter points around the spiral with randomly offsets
zdata = 15 * np.random.random(100)
xdata = np.sin(zdata) + 0.1 * np.random.randn(100)
ydata = np.cos(zdata) + 0.1 * np.random.randn(100)
ax.scatter(xdata, ydata, zdata, color="grey")

plt.show()
```

Contour plots
Use `plot_wireframe` for wireframe plots, and `plot_surface` for surface plots.

```python
def himmelblau(x, y):
    return (x ** 2 + y - 11) ** 2 + (x + y ** 2 - 7) ** 2
of.fig3()
op.surface(himmelblau, (-5,5), (-5,5), alpha=0.5)
plt.show()
```

```python
def himmelblau(x, y):
    return (x ** 2 + y - 11) ** 2 + (x + y ** 2 - 7) ** 2
of.fig3()
op.wireframe(himmelblau, (-5,5), (-5,5), alpha=0.5)
plt.show()
```

```python
of.fig3()
op.contour(himmelblau, (-5,5), (-5,5), levels=100, alpha=0.5)
plt.show()
```
